```python
#Import Open CV
import cv2
from google.colab.patches import cv2_imshow
#membaca citra dari file
image = cv2.imread('Spidey.jpeg')
cv2_imshow(image)
cv2.waitKey(0)
```

```python
#mengecek ukuran pixel
if image is not None:
  tinggi, lebar, channel = image.shape
  print(f'Tinggi: {tinggi}, Lebar: {lebar}, Channel: {channel}')
else:
  print('Gagal membaca citra')
```

```python
#mengubah ke Grayscale
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
cv2_imshow(gray_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
cv2_imshow(gray_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2LAB)
cv2_imshow(gray_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2LUV)
cv2_imshow(gray_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
#resize pixel
resized_image = cv2.resize(gray_image, (200, 300))
cv2_imshow(resized_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
cv2.imwrite('latihan_citra_digital.jpg', resized_image)
```
