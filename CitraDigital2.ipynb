```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

gambar = cv2.imread ('Ghiss.jpeg')
cv2_imshow(gambar)
cv2.waitKey(0)
```

```python
#Konversi Grayscale

gray = cv2.cvtColor(gambar, cv2.COLOR_BGR2GRAY)
plt.title('Konversi Grayscale')
plt.imshow(gray, cmap='gray')
```

```python
#Hitung histogram grayscale
histogram_gray = cv2.calcHist([gray], [0], None, [256], [0,256])

#Plot Histogram
plt.plot(histogram_gray, color='black')
plt.title("Histogram Grayscale")
plt.xlabel("Intensitas (0 - 255)")
plt.ylabel("Jumlah Piksel")
plt.show()
```

```python
gambar_rgb = cv2.cvtColor(gambar, cv2.COLOR_BGR2RGB)

plt.title('konversi RGB')
plt.imshow(gambar_rgb)
```

```python
# Hitung histogram untuk channel Red
hist_red = cv2.calcHist([gambar_rgb], [2], None, [256], [0,256])

# Plot histogram Red
plt.plot(hist_red, color='r')
plt.title("Histogram Red Channel")
plt.xlabel("Intensitas (0-255)")
plt.ylabel("Jumlah Piksel")
plt.show()
```

```python
# Hitung histogram untuk channel green
hist_green = cv2.calcHist([gambar_rgb], [1], None, [256], [0,256])

# Plot histogram Green
plt.plot(hist_green, color='g')
plt.title("Histogram Green Channel")
plt.xlabel("Intensitas (0-255)")
plt.ylabel("Jumlah Piksel")
plt.show()
```

```python
# Hitung histogram untuk channel blue
hist_blue = cv2.calcHist([gambar_rgb], [0], None, [256], [0,256])

# Plot histogram Blue
plt.plot(hist_blue, color='b')
plt.title("Histogram Blue Channel")
plt.xlabel("Intensitas (0-255)")
plt.ylabel("Jumlah Piksel")
plt.show()
```

```python
# Warna channel
color = ('b', 'g', 'r')

plt.figure(figsize=(8,6))
for i, col in enumerate(color):
    hist = cv2.calcHist([gambar_rgb], [i], None, [256], [0,256])
    plt.plot(hist, color=col, label=col.upper())
    plt.xlim([0,256])

plt.title("Histogram Gabungan RGB")
plt.xlabel("Intensitas (0-255)")
plt.ylabel("Jumlah Piksel")
plt.legend()
plt.show()
```

```python
# Tambah kecerahan (misal +100)
img_bright = cv2.convertScaleAbs(gambar, alpha=1, beta=100)

# Tampilkan sebelum & sesudah
plt.subplot(1,2,1)
plt.imshow(cv2.cvtColor(gambar, cv2.COLOR_BGR2RGB))
plt.title("Asli")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(cv2.cvtColor(img_bright, cv2.COLOR_BGR2RGB))
plt.title("Lebih Cerah (+100)")
plt.axis("off")

plt.show()
```
